```python
import pandas as pd

df = pd.read_csv("/kaggle/input/datasets/sahilislam007/sales-dataset/Sales Dataset.csv")
```

```python
df.groupby("Product Category")["Quantity"].sum()
```

```python
import matplotlib.pyplot as plt
```

```python
id="e9m1qa"
sales_by_category = df.groupby("Product Category")["Quantity"].sum()

sales_by_category.plot(kind="bar")

plt.xlabel("Product Category")
plt.ylabel("Quantity Sold")
plt.title("Quantity Sold by Product Category")

plt.show()
```

```python
daily_sales = df.groupby("Date")["Total Amount"].sum()

daily_sales.head(10)
```

```python
daily_sales.plot(figsize = (12,5))

plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.title("Daily Sales Trend")

plt.show()
```

```python
df["Gender"].value_counts()
```

```python
df["Gender"].value_counts().plot(kind = "bar")

plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.title("Customer Distribution by Gender")
plt.show()
```

```python
age_sales = df.groupby("Age")["Total Amount"].sum()

age_sales
```

```python
age_sales.idxmax(), age_sales.max()
```

```python
age_sales.plot(kind="bar", figsize=(12, 5))

plt.xlabel("Age")
plt.ylabel("Total Sales")
plt.title("Total Sales by Age")

plt.show()
```
